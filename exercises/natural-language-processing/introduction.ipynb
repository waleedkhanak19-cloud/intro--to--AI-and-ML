```python
import nltk  as NLTK
import spacy as spaCy
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.probability import FreqDist
from nltk.util import ngrams
from nltk import pos_tag
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
file_path = "/content/harry_potter.txt"
with open(file_path, "r", encoding="utf-8") as file:
    harry_potter_text = file.read()
print(len(harry_potter_text))
```

```python
# NLTK sent_tokenize() divides a text into sentences.
sentences = sent_tokenize(harry_potter_text)

# NLTK word_tokenize() divides a text into words.
words = word_tokenize(harry_potter_text)

print("\nNumber of sentences:", len(sentences))

print("\nFirst 5 sentences:")
for sentence in sentences[:5]:
    print(sentence)
```

```python
# word_tokenize() divides the text into individual tokens.
tokens = word_tokenize(harry_potter_text)

print("\nNumber of tokens:", len(tokens))


# Convert tokens to lowercase so that "Harry" and "harry"
# are counted as the same token.
tokens_lower = [token.lower() for token in tokens]


# FreqDist calculates how often every token appears.
frequency = FreqDist(tokens_lower)

print("\n20 most frequent tokens:")

for token, count in frequency.most_common(20):
    print(token, ":", count)


# We can access the frequency of a specific word like this:
print("\nFrequency of 'harry':", frequency["harry"])
```

```python
# word_tokenize() divides the text into individual tokens.
tokens = word_tokenize(harry_potter_text)

print("\nNumber of tokens:", len(tokens))


# Convert tokens to lowercase so that "Harry" and "harry"
# are counted as the same token.
tokens_lower = [token.lower() for token in tokens]


# FreqDist calculates how often every token appears.
frequency = FreqDist(tokens_lower)

print("\n20 most frequent tokens:")

for token, count in frequency.most_common(20):
    print(token, ":", count)


# We can access the frequency of a specific word like this:
print("\nFrequency of 'harry':", frequency["harry"])

```

```python
# Here we automatically choose a reasonably short sentence
# containing the word "Harry".
chosen_sentence = None

for sentence in sentences:
    sentence_tokens = word_tokenize(sentence)

    if "Harry" in sentence and 5 <= len(sentence_tokens) <= 15:
        chosen_sentence = sentence
        break


print("\nChosen sentence:")
print(chosen_sentence)


# Tokenize the chosen sentence.
chosen_tokens = word_tokenize(chosen_sentence)

print("\nTokens:")
print(chosen_tokens)
```

```python
#An n-gram is a sequence of n consecutive tokens.
#
# unigram = 1 token
# bigram  = 2 tokens
# trigram = 3 tokens
# etc.

print("\nALL N-GRAMS:")

for n in range(1, len(chosen_tokens) + 1):

    current_ngrams = list(ngrams(chosen_tokens, n))

    print(f"\n{n}-grams:")

    for gram in current_ngrams:
        print(gram)


# We can also create only bigrams, for example:
bigrams = list(ngrams(chosen_tokens, 2))

print("\nBigrams:")
print(bigrams)
```

```python
# pos_tag() assigns a Part-of-Speech tag to each token.
#
# Examples:
# NN  = noun
# NNP = proper noun
# VB  = verb
# JJ  = adjective
# RB  = adverb

nltk_pos_tags = pos_tag(chosen_tokens)

print("\nNLTK POS tags:")

for word, tag in nltk_pos_tags:
    print(word, "->", tag)
```
